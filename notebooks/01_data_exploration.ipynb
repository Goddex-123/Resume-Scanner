# 01. Data Exploration

This notebook explores the synthetic dataset generated for training the ML Resume Scanner.
It covers class distribution, lengths, and skill frequency.

```
import os
import sys
sys.path.append('..')

import pandas as pd
import plotly.express as px
from resume_scanner.ml.dataset import ResumeJDDatasetGenerator

# Generate dataset
gen = ResumeJDDatasetGenerator(seed=42)
entries = gen.generate(pairs_per_domain=100)
df = pd.DataFrame([vars(e) for e in entries])
df.head()
```

## Class Distribution

```
fig = px.histogram(df, x='match_label', color='domain', title='Class Distribution by Domain', barmode='group')
fig.show()
```

## Text Length Analysis

```
df['resume_len'] = df['resume_text'].apply(len)
df['jd_len'] = df['job_description'].apply(len)

fig = px.scatter(df, x='resume_len', y='jd_len', color='match_label', title='Resume vs JD Length')
fig.show()
```
