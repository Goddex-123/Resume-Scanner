# 02. Feature Engineering

Extracts lexical and semantic features from the resume-JD pairs.

```
import os
import sys
sys.path.append('..')

import pandas as pd
import plotly.express as px
from resume_scanner.ml.dataset import ResumeJDDatasetGenerator
from resume_scanner.ml.features import FeatureExtractor
from resume_scanner.ml.embeddings import get_encoder

# Generate a small sample
gen = ResumeJDDatasetGenerator(seed=42)
entries = gen.generate(pairs_per_domain=20)
df = pd.DataFrame([vars(e) for e in entries])

print('Initializing Feature Extractor (loading Sentence-BERT)...')
encoder = get_encoder()
extractor = FeatureExtractor(encoder=encoder)
```

## Extracting Features

```
features_list = []
for i, row in df.iterrows():
    feats = extractor.extract(row['resume_text'], row['job_description'])
    feats['match_label'] = row['match_label']
    feats['domain'] = row['domain']
    features_list.append(feats)

feat_df = pd.DataFrame(features_list)
feat_df.head()
```

## Feature Correlations

```
corr = feat_df.select_dtypes('number').corr()
fig = px.imshow(corr, title='Feature Correlation Matrix')
fig.show()
```
